**2. Базові операції**

Створіть Numpy-масив зі значеннями [12, 24, 36, 48, 60].

Виконайте наступні операції над **оригінальним** масивом і запишіть у нову змінну результат (кожен пункт оцінюється в 0.5 бала):
- Помножте всі елементи на 2
- Додайте до всіх елементів 10
- Піднесіть всі елементи до квадрату
- Обчисліть квадратний корінь з усіх елементів

Виведіть результат виконання кожної операції.

```python
# Create a NumPy array
base_values = np.array([12, 24, 36, 48, 60])

# Apply element operations
doubled_values = base_values * 2      # Multiply each element by 2
values_plus_ten = base_values + 10    # Add 10 to each element
squared_values = base_values ** 2     # Square each element
square_root_values = np.sqrt(base_values)  # Compute square root of each element

# Display the results
display(doubled_values)
display(values_plus_ten)
display(squared_values)
display(square_root_values)
```

**3. Статистичні функції**

Створіть Numpy-масив `data` зі значеннями [18, 27, 33, 41, 52, 68, 79].

Обчисліть використовуючи функціонал numpy (кожен пункт оцінюється в 0.5 бала):
- Середнє значення
- Мінімум
- Максимум
- Суму всіх елементів

```python
# Create a NumPy array
data = np.array([18, 27, 33, 41, 52, 68, 79])

# Calculate summary statistics
mean_data = data.mean()
min_data = data.min()
max_data = data.max()
sum_data = data.sum()

# Display the results
display(mean_data)
display(min_data)
display(max_data)
display(sum_data)
```

**4. Індексація**

Створіть Numpy-масив `arr` зі значеннями [15, 22, 35, 44, 58, 67].

Виведіть (кожен пункт оцінюється в 0.5 бала):
- Перший елемент
- Останній елемент
- Елементи з індексами 1 до 3
- Елементи більші за 40

```python
# Create a NumPy array
arr = np.array([15, 22, 35, 44, 58, 67])

# Access elements by position
first_element = arr[0]
last_element = arr[-1]

# Slice a subset of elements
elements = arr[1:4]

# Filter elements greater than 40
elements_more_40 = arr[arr > 40]

# Display the results
display(first_element)
display(last_element)
display(elements)
display(elements_more_40)
```
