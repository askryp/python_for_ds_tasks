**1. Функція з параметром**

Створіть функцію `show_menu()`, яка приймає перелік страв, наприклад ['Піца', 'Паста', 'Салат'] і виводить меню ресторану в наступному вигляді:
```
=== МЕНЮ ===
1. Піца
2. Паста
3. Салат
=============
```

Викличте цю функцію на списку ["Суп грибний", "Котлета по-київськи", "Лимонад"].

````python
# Function to display menu items with formatting
# Parameter: menu (list of strings representing menu items)
def show_menu(menu):
    # Print the menu header
    print("=== МЕНЮ ===")
    # Iterate through each item with its index (enumerate gives index and item)
    # i is the index, item is the menu item string
    for i, item in enumerate(menu):
        # Print each item numbered starting from 1 (i+1 because enumerate starts at 0)
        print(f"{i+1}. {item}")
    # Print the menu footer
    print("=============")
````

````python
# Create a list of available dishes for the menu
menu_items = ["Суп грибний", "Котлета по-київськи", "Лимонад"]

# Call the show_menu function
show_menu(menu_items)
````

**2. Функція з параметром**

Створіть функцію `print_square(n)`, яка виводить квадрат числа.

Наприклад: `print_square(5)` → "Квадрат числа 5 = 25"

Викличте її з різними числами.

````python
# Function to calculate and display the square of a number
# Parameter: n (numeric value to square)
def print_square(n):
    # Calculate the square of the input number
    squared_number = n ** 2
    # Print the result in a formatted string
    print(f"Квадрат числа {n} = {squared_number}")
````

````python
print_square(1)
print_square(25)
print_square(10)
````

**3. Функція з поверненням значення**

Створіть функцію `multiply(a, b)`, яка повертає добуток двох чисел.

Викличте функцію на власноруч заданих значеннях та виведіть результат.


````python
# Function to multiply two numbers and display the result
# Parameters: a, b (numeric values to multiply)
def multiply(a, b):
    multiplied_number = a * b
    print(f"Добуток {a} та {b} дорівнює {multiplied_number}")
````

````python
# Test the multiply function with arguments 1 and 4
multiply(1, 4)
````

**4. Функція пошуку максимумального значення**

Створіть функцію `find_max(numbers)`, яка:
- Приймає список чисел
- Повертає найбільше число в списку

Протестуйте на списку `[15, 42, 8, 23, 56, 19]`.


````python
# Function to find and display the maximum value in a list
# Parameter: numbers (list of numeric values)
def find_max(numbers):
    max_number = max(numbers)
    print(max_number)
````

````python
number_list = [15, 42, 8, 23, 56, 19]


# Find and display the maximum value from the list
find_max(number_list)
````

**5. Функція з умовами**

Створіть функцію `check_grade(grade)`, яка:
- Приймає оцінку (число)
- Повертає "Здав" якщо grade >= 60
- Повертає "Не здав" якщо grade < 60

Протестуйте на різних оцінках.


````python
# Function to check if a grade is passing or failing
# Parameter: grade (numeric value representing a grade)
def check_grade(grade):
    # If grade is 60 or higher, the student passed
    if grade >= 60:
        print("Здав")
    # Otherwise, the student failed
    else:
        print("Не здав")
````

````python
grade_1 = 60
grade_2 = 26
grade_3 = 99
grade_4 = 87

# Test the function with all grade values
check_grade(grade_1)
check_grade(grade_2)
check_grade(grade_3)
check_grade(grade_4)
````
