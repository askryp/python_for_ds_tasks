## 1️⃣ Встановлення залежностей

```python
import subprocess
import sys

# Встановлення необхідних пакетів
packages = ['numpy', 'pandas', 'matplotlib', 'scikit-learn', 'statsmodels', 'sympy']

for package in packages:
    try:
        __import__(package)
        print(f"✓ {package} уже встановлений")
    except ImportError:
        print(f"Встановлюю {package}...")
        subprocess.check_call([sys.executable, "-m", "pip", "install", package, "--quiet"])
```
