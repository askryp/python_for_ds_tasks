**1. Цикл while**

Задано змінну `counter`, яка рівна -3.

Напишіть програму, використовуючи `while`, яка в циклі додає до змінної `counter` +1 і зупиняється, коли змінна `counter` стане суворо більше за 5. На кожній ітерації циклу виводьте значення змінної `counter`.



```python
# Initialize counter variable to -3
counter = -3

# Loop while counter
while counter < 5:
    # Increment counter by 1 on each iteration
    counter += 1
    # Print the current value of counter
    print(f"Лічильник:{counter}")
```

**2. List comprehension - прості операції**

Дано список чисел:

```python
numbers = [11, 4, 9, 16, 7, 5]
```


Створіть нові списки використовуючи list comprehension, які містять (кожен пункт оцінюється в 0.5 бала):
- Квадрати всіх чисел
- Тільки парні числа
- Всі числа помножені на 10

Виведіть новоутворені списки.

```python
# Create a list of squared numbers using list comprehension
numbers_squared_list = [x**2 for x in numbers]

# Create a list containing only even numbers
even_numbers_list = [x for x in numbers if x % 2 == 0]

# Create a list by multiplying each number by 10
numbers_times_ten_list = [x * 10 for x in numbers]

print(f"Квадрати всіх чисел: {numbers_squared_list}")
print(f"Тільки парні числа: {even_numbers_list}")
print(f"Всі числа помножені на 10: {numbers_times_ten_list}")
```

**3. List comprehension - комбінована логіка**

Дано список чисел від 1 до 10. Створіть новий список, де виконується наступна логіка:
- Якщо число парне - помножте його на 2
- Якщо непарне - віднімніть 1

Виведіть результуючий список на екран.

```python
numbers = list(range(1, 11))
```

```python
# Create a new list using conditional list comprehension:
transformed_numbers_list = [x * 2 if x % 2 == 0 else x - 1 for x in numbers]

print(f"Transformed list: {transformed_numbers_list}")
```
