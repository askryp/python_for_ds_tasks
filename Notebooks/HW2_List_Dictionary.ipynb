**1. Списки - базові операції**

Створіть список температур за тиждень (у градусах Цельсія) з назвою
`temperatures` і значеннями [22, 25, 19, 23, 26, 20, 21, 22].


Виконайте (кожен пункт оцінюється в 0.5 бала):
- Виведіть перший елемент
- Виведіть останній елемент
- Додайте температуру 24
- Видаліть температуру 19
- Виведіть довжину списку
- Обчисліть суму всіх температур
- Знайдіть максимальну температуру

```python
# Define a list of temperatures
temperatures = [22, 25, 19, 23, 26, 20, 21, 22]

# First element
print(f"First element: {temperatures[0]}")

# Last element
print(f"Last element: {temperatures[-1]}")
# print(f"Last element: {temperatures[7]}") # This also works but is less flexible

#Add element
temperatures.append(24)
print(f"Add element: {temperatures}")

# Delete element
del temperatures[2]  # Remove the element at index 2 (19)
# temperatures.remove(19)  # Remove the first occurrence of 19
print(f"After deletion: {temperatures}")

# How many elements
print(f"Number of elements: {len(temperatures)}")

# Sum of elements
print(f"Sum of elements: {sum(temperatures)}")

# Max temperature
print(f"Max temperature: {max(temperatures)}")
```


**2. Словники - базові операції**

Створіть словник про себе, який містить ключі `name`, `age`, `city` і значення, які відповідають вашему імені, віку і місту (можна вставити будь-які значення, не обовʼязково свої).


Виконайте (кожен пункт оцінюється в 0.5 бала):
- Виведіть значення ключа "name"
- Змініть значення ключа "age" на 21
- Додайте новий ключ "hobby" зі значенням "programming"
- Виведіть всі ключі
- Виведіть всі значення


```python
# Define a dictionary
student = {
    "name": "Olena",
    "age": 33,
    "city": "Kyiv"
}

# Show the key name
print(f"Key 'name': {student['name']}")

# Update the age
student['age'] = 21
print(f"Updated age: {student['age']}")

# Added a new key 'hobby'
student['hobby'] = 'programming'
print(f"Added 'hobby' key: {student['hobby']}")

# Show all keys
print(f"All keys: {student.keys()}")

# Show all values
print(f"All values: {student.values()}")
```
