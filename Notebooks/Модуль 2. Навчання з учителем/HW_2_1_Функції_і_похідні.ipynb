1. Задана функція залежності оцінки від кількості годин, витрачених на навчання: $ f(x) = 10x - 5 $.
  1. З допомогою Python обчисліть похідну цієї функції.
  2. Використовуйте бібліотеку `matplotlib` для побудови графіків функції та її похідної в діапазоні значень від 0 до 10 годин. Проаналізуйте, як зміна кількості годин вивчення впливає на оцінку.



```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Визначення діапазону від 0 до 10 годин
x = np.linspace(0, 10, 100)

# Розрахунок функції та її похідної
f = 10 * x - 5
df = np.full_like(x, 10) # Похідна f'(x) = 10

# Створення графіку
plt.figure(figsize=(10, 5))

# Малюємо тільки функцію та її похідну
plt.plot(x, f, label='f(x) = 10x - 5 (Оцінка)', color='blue', linewidth=2)
plt.plot(x, df, label="f'(x) = 10 (Швидкість росту)", color='green', linestyle='--', linewidth=2)

# Додавання легенди, міток відповідно до умов задачі
plt.title("Залежність оцінки від кількості годин навчання та її похідна")
plt.xlabel("Кількість годин навчання (x)")
plt.ylabel("Оцінка (y)")
plt.legend()
plt.grid(True)

plt.show()
```

На графіку видно пряму лінійну залежність. Чим більше годин учень приділяє навчанню, то вищу оцінку він отримує.

2. Використовуючи Python і `SymPy`, обчисліть похідні для функцій $ f(x) = \sin(x) $ та $ f(x) = \ln(x) $. Зобразіть ці функції та їхні похідні за допомогою `matplotlib`.

```python
import sympy as sp
```

```python
# Ініціалізація символьної змінної
x = sp.Symbol('x')

# Обчислення похідної для тригонометричної функції
f_sin = sp.sin(x)
derivative_f_sin = sp.diff(f_sin, x)

# Обчислення похідної для логарифмічної функції
f_ln = sp.log(x)
derivative_f_ln = sp.diff(f_ln, x)

# Виведення результатів
print(f"Похідна від f(x) = sin(x) дорівнює:  f'(x) = {derivative_f_sin}")
print(f"Похідна від f(x) = ln(x)  дорівнює:  f'(x) = {derivative_f_ln}")
```

```python
# Визначення діапазону аргументу від 0 до 20
x = np.linspace(0, 20, 100)

# Рахуємо значення тригонометричної функції та її похідної
f_sin = np.sin(x)
df_sin = np.cos(x)

# Створення графіку
plt.figure(figsize=(10, 6))
plt.plot(x, f_sin, label='f(x) = sin(x)', color='blue', linewidth=2)
plt.plot(x, df_sin, label="f'(x) = cos(x)", color='green', linestyle='--', linewidth=2)

# Підписи та заголовок
plt.title("Графік функції f(x) = sin(x) та її похідної f'(x) = cos(x)")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.grid(True)

plt.show()
```

```python
# Визначення діапазону аргументу від 1 до 20
x = np.linspace(1, 20, 100)

# Рахуємо значення логарифмічної функції та її похідної
f_ln = np.log(x)
df_ln = 1/x

# Створення графіку
plt.figure(figsize=(10, 6))
plt.plot(x, f_ln, label='f(x) = ln(x)', color='blue', linewidth=2)
plt.plot(x, df_ln, label="f'(x) = 1/x", color='green', linestyle='--', linewidth=2)

# Підписи та заголовок
plt.title("Графік функції f(x) = ln(x) та її похідної f'(x) = 1/x")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.grid(True)

plt.show()
```


3. Загальні витрати компанії від обсягу виробництва задані функцією $ f(x) = x^2 - 5x + 1 $, де $ x $ — кількість вироблених одиниць. Обчисліть похідну цієї функції з допомогою SymPy та візуалізуйте значення функції і її похідної на відрізку від 0 до 100.

```python
# Визначаємо символічну змінну
x = sp.Symbol('x')

# Лінійна функція
f_lin = x ** 2 - 5 * x + 1
derivative_f_lin = sp.diff(f_lin, x)
print(derivative_f_lin)
```

```python
# Визначення діапазону від 0 до 100 вироблених одиниць
x = np.linspace(0, 100, 100)

# Рахуємо значення функції та її похідної для кожного x
f_vals = x**2 - 5*x + 1
df_vals = 2 * x - 5  # Формула, яку нам знайшов SymPy (2*x - 5)

# Створення графіку
plt.figure(figsize=(10, 6))
plt.plot(x, f_vals, label='f(x) = x² - 5x + 1 (Загальні витрати)', color='blue', linewidth=2)
plt.plot(x, df_vals, label="f'(x) = 2x - 5 (Граничні витрати)", color='green', linestyle='--', linewidth=2)

# Підписи та заголовок
plt.title("Залежність загальних витрат компанії від обсягу виробництва та їхня похідна")
plt.xlabel("Кількість вироблених одиниць (x)")
plt.ylabel("Витрати (y)")
plt.legend()
plt.grid(True)

plt.show()
```
